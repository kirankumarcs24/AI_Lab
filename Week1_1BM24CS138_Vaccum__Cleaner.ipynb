```python
def vacuum_cleaner(location, status):
    if status == "Dirty":
        print("Room", location, "is Dirty")
        print("Action: Suck")
        status = "Clean"

    if location == "A":
        print("Action: Move Right")
        location = "B"
    else:
        print("Action: Move Left")
        location = "A"

    return location, status


# Initial state
location = "A"
room_A = "Dirty"
room_B = "Dirty"

print("Initial State:")
print("A:", room_A)
print("B:", room_B)
print("Vacuum:", location)

# Clean Room A
location, room_A = vacuum_cleaner(location, room_A)

# Clean Room B
location, room_B = vacuum_cleaner(location, room_B)

print("\nFinal State:")
print("A:", room_A)
print("B:", room_B)
```
